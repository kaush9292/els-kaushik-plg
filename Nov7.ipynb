```python
# read the file from dbfs path "dbfs:/FileStore/tables/journal_dummy_dataset.csv"
df = spark.read.csv("dbfs:/FileStore/tables/journal_dummy_dataset.csv", header=True, inferSchema=True)
display(df)
```

```python
# create a new column 'Publication_Year' and split the year alone from the 'Publication_Date' column.
from pyspark.sql.functions import year, col

df = df.withColumn('Publication_Year', year(col('Publication Date')))
display(df)

```
